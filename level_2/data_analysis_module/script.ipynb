```python
check_unique_columns= ['country', 'customer_type', 'deposit_type']

for col in check_unique_columns:
    unique_column = hotel_df[col].unique()
    len_unique_column = len(unique_column)
    print(f"{col} has {len_unique_column} unique values. \n {unique_column}")
```

```python
country_count = hotel_df['country'].value_counts().to_dict()
country_list = [country for country, val in country_count.items() if val > 1000]
country_list
```

```python
hotel_df_subset = hotel_df.loc[
    hotel_df['country'].isin(country_list)
    &  hotel_df['is_canceled']==0
    ]
```

```python
hotel_df_subset.loc['reservation_status_date'] = pd.to_datetime(hotel_df_subset['reservation_status_date'])
```

```python
hotel_df_subset.loc['reservation_status_year'] = hotel_df_subset['reservation_status_date'].dt.year
hotel_df_subset.head()
```

```python
# hotel_df_subset.groupby('country').size()
grouped_hotel_df_subset = hotel_df_subset.groupby(['country','reservation_status_year']).size().reset_index(name='count_reservation_per_year')
grouped_hotel_df_subset
```

```python
grouped_hotel_df_subset.sort_values(['count_reservation_per_year', 'reservation_status_year'], ascending=False)

# Check
# hotel_df_subset.loc[ (hotel_df_subset['country']=='PRT')
#  & (hotel_df_subset['reservation_status_year']==2016)]
```
